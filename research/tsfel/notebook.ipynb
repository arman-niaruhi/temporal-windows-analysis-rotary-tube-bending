```python
dfs = pd.read_csv("../data/features.csv", index_col = "Time_[s]")
targets = pd.read_csv("../data/targets.csv", index_col = False)
df = dfs[dfs["Experiment_ID"]==2].drop(columns=["Experiment_ID"])
df
```

```python
import pandas as pd
import tsfel
import matplotlib.pyplot as plt
import seaborn as sns

# TSFEL configuration: all domains (statistical, temporal, spectral)
cfg = tsfel.get_features_by_domain()

# Sampling frequency (based on 0.05 s interval)
fs = 1 / 0.05  # 20 Hz

# Extract features
features = tsfel.time_series_features_extractor(cfg, df, fs=fs)
features
```

Feature and	Meaning
### Absolute energy	Sum of squared values of the signal → measures signal strength.
### Area under the curve	Integral of the signal over time → total “movement” magnitude.
### Autocorrelation	How similar the signal is to a delayed version of itself → periodicity or repeating patterns.
### Average power	Mean power of the signal → similar to energy but normalized over time.
### Centroid	Weighted average of signal positions → shows “center of mass” of the signal in time or frequency domain.
### ECDF Percentile Count_X	Number of values falling into the Xth percentile of the empirical cumulative distribution → measures distribution.
### ECDF Percentile_X	The value at the Xth percentile → gives statistical quantiles.
### Wavelet variance_YHz	Variance of the signal at frequency YHz using wavelet decomposition → frequency-specific energy.
### Zero crossing rate	Number of times the signal crosses zero → indicates oscillations or vibration.
